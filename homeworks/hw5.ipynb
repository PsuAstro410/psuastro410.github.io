## Lagrange points of the circular restricted three-body problem

We assume two bodies with mass $M_1$ and $M_2$ orbit their common center of mass on a circular orbit with separation $a$. A third body (spacecraft, asteroid, dust grain, etc.) is assumed to be so light that it feels the gravity of the other two but does not affect them. This is the Circular Restricted 3-Body Problem (CR3BP). In this problem, we will calculate the so-called Lagrange points of the CR3BP, where the force between the two massive bodies cancels, and the third body can co-orbit with the two. For instance, JWST is located at second Earth-Sun Lagrange point $L_2$.

It is easiest to work in the frame that co-rotates with the orbiting $M_1$ and $M_2$, because in that frame the two massive bodies sit still. The orbital angular frequency follows from Kepler's third law,

$$\Omega = \sqrt{\frac{G(M_1 + M_2)}{a^3}}.$$

In this exercise, we will adopt units in which

$$G = M_1 + M_2 = a = 1$$

so that $\Omega = 1$, and define the the mass ratio

$$\mu \equiv \frac{M_2}{M_1 + M_2}$$

Putting the center of mass at the origin and the two bodies on the $x$-axis, the primary (mass $1-\mu$) sits at $x_1 = -\mu$ and the secondary (mass $\mu$) sits at $x_2 = 1-\mu$. The distances from a test particle at $(x, y)$ in the orbital plane to each body are

$$r_1 = \sqrt{(x+\mu)^2 + y^2}, \qquad r_2 = \sqrt{(x-1+\mu)^2 + y^2}.$$

In the rotating frame the test particle feels gravity from both bodies plus the centrifugal force. We fold both forces into a single effective potential $\Phi_{\rm eff}$. The gravitational potential per unit mass from the two bodies is $-GM_1/r_1 - GM_2/r_2$, which in our units is $-(1-\mu)/r_1 - \mu/r_2$. The centrifugal force on a particle at cylindrical radius $R = \sqrt{x^2+y^2}$ from the rotation axis is $\mathbf{F}_{\rm cen} = \Omega^2 R\,\hat{\mathbf{R}}$ per unit mass, directed outward. Writing $\mathbf{F}_{\rm cen} = -\nabla \Phi_{\rm cen}$ gives $\Phi_{\rm cen} = -\tfrac{1}{2}\Omega^2 R^2$.

Adding the two pieces, and setting $\Omega = 1$:

$$
\Phi_{\rm eff}(x, y) = -\frac{1-\mu}{\sqrt{(x+\mu)^2+y^2}} \; - \; \frac{\mu}{\sqrt{(x-1+\mu)^2+y^2}} \; - \; \frac{1}{2}\left(x^2 + y^2\right)\\
= -\frac{1-\mu}{r_1} - \frac{\mu}{r_2} - \frac{1}{2} R^2.
$$



Throughout this problem set we will use

$$\mu = 0.3$$

i.e. a binary with $M_1 = 0.7$ and $M_2 = 0.3$ in units of the total mass. This is a deliberately large mass ratio: it is far from the Sun–Earth ($\mu \approx 3\times10^{-6}$) or Earth–Moon ($\mu \approx 0.012$) values, but it spreads the five equilibrium points out so that they are easy to see.
