As a reminder, the 4th order Runge-Kutta integrator solves an equation of the form
$$
\frac {d\vec{y}}{dt} = \vec{f}(t, \vec{y}),
$$
for a function $\vec{y}$ of time $t$. It does splitting the times into steps $t_n$ with spacing $\Delta t$, with function values $\vec{y}_n \equiv \vec{y}(t_n)$. It advances to $\vec{y}_{n+1}$ using the value of $\vec{y}_n$ via
\begin{eqnarray}
 \vec{k}_1 & = & \vec{f} ( t_n, \vec{y}_n), \\
 \vec{k}_2 & = & \vec{f} \left(t_n + \frac{\Delta t}{2} , \vec{y}_n + \frac{\Delta t}{2} \cdot \vec{k}_1 \right), \\
 \vec{k}_3 & = & \vec{f} \left(t_n + \frac{\Delta t}{2} , \vec{y}_n + \frac{\Delta t}{2} \cdot \vec{k}_2\right), \\
 \vec{k}_4 & = & \vec{f} \left(t_n + \Delta t , \vec{y}_n + \Delta t \cdot \vec{k}_3 \right),
\end{eqnarray}
with
$$
\vec{y}_{n+1} = \vec{y}_n + \frac{\Delta t}{6}\left(\vec{k}_1 + 2\vec{k}_2 + 2\vec{k}_3 + \vec{k}_4 \right).
$$

# Problem 2 - Double Pendulum Integration (4 pt)

The double pendulum is a famously chaotic system.  A description of this system can be found See the [wikipedia page on double pendulum](https://en.wikipedia.org/wiki/Double_pendulum).

The equations of motion for a double pendulum consisting of two rigid rods of equal length $l$ and mass $m$ are:
    
$$\dot{\theta}_1 = \frac{6}{I} \frac{2 {p_\theta}_1 - 3 \cos(\theta_1 - \theta_2) {p_\theta}_2}{16 - 9 \cos^2 (\theta_1 - \theta_2)}$$
    
$$\dot{\theta}_2 = \frac{6}{I} \frac{8 {p_\theta}_2 - 3 \cos(\theta_1 - \theta_2) {p_\theta}_1}{16 - 9 \cos^2 (\theta_1 - \theta_2)}$$
    
$${\dot{p}_\theta}_1 = -\frac{1}{2} I \left (\dot{\theta}_1\dot{\theta}_2 \sin(\theta_1 - \theta_2) + 3 \frac{g}{l} \sin \theta_1 \right )$$
    
$${\dot{p}_\theta}_2 = -\frac{1}{2} I \left (-\dot{\theta}_1\dot{\theta}_2 \sin(\theta_1 - \theta_2) +  \frac{g}{l} \sin \theta_2 \right )$$
    
where $I = m l^2$ is the moment of inertia-like term, $g$ is the acceleration due to gravity, $\theta_1$ and $\theta_2$ are
the angles of the two rods with respect to the vertical, and $p_{\theta 1}$
and $p_{\theta 2}$ are generalized momenta:

# Problem 3 - Lane-Emden Equation (3 pt)

The equations of stellar structure are perhaps the most important boundary value problem in astrophysics. The first is the continuity equation, describing how the total mass $m$ enclosed within a radius $r$ depends on density $\rho(r)$, or
$$
\frac{dm}{dr} = 4\pi r^2 \rho.
$$
The second is the equation of hydrostatic equilibrium, describing the balance between pressure $P(r)$ and gravity,
$$
\frac{1}{\rho} \frac{d P}{d r} = - \frac{G m}{r^2}.
$$

These equations are supplemented by an energy equation and an equation of state, that codes like MESA solve to calculate the evolution of stellar structure with time. These equations can be simplified greatly by assuming that pressure is related to density via a polytrope, or
$$
P = K \rho^{1 + \frac{1}{n}},
$$
where $n$ is the polytropic index, and $K$ is a constant of proportionality. If we re-write $\rho$ in terms of the core density $\rho_0$ and a dimensionless function $\theta$ via
$$
\rho = \rho_0 \theta^n,
$$
and the radius $r$ in terms of a dimensionless variable $\xi$ via
$$
r = \alpha \xi,
$$
the above two equations can be combined into the [Lane-Emden equation](https://en.wikipedia.org/wiki/Lane%E2%80%93Emden_equation)
$$
\frac{1}{\xi^2} \frac{d}{d \xi} \left( \xi^2 \frac{d \theta}{d \xi} \right) + \theta^n = 0.
$$
Here, the constant $\alpha$ is
$$
\alpha^2 = \frac{(n+1) K \rho_0^{ \frac{1}{n} - 1 }}{4\pi G}.
$$

At the stellar surface $r = \alpha \xi_1$, the density vanishes, and
$$
\theta(\xi_1) = 0.
$$
At the core, the gravity $(G m/r^2)_{r\to 0} = 0$, and so
$$
\frac{d \theta}{d \xi} \bigg|_{\xi_0} = 0,
$$
for $\xi_0 = 0$. By definition, the core density $\rho(0) = \rho_0$, and so
$$
\theta(\xi_0) = 1.
$$
The problem is to solve for $\xi_1$. We will calculate where $\theta(\xi)<\epsilon$ to calculate $\xi_1$, which we will do via adaptive timestepping.

The solution to these two equations satisfy the boundary conditions
$$
\theta(\xi_0) = 1 \qquad \psi(\xi_0) = 0.
$$
However, the computer has trouble literally evaluating equation (1) when $\xi \to 0$. Instead, we will apply the boundary conditions to an inner boundary $\xi_0 \ll 1$, but not quite $\xi_0 = 0$. Specifically, we expand $\theta$ and $\psi$ near $\xi = 0$ as
\begin{align}
\theta(\xi) &= 1 + \frac{\eta}{2} \xi^2 + \mathcal{O}(\xi^4) \\
\psi(\xi) &= \eta \xi + \mathcal{O}(\xi^3)
\end{align}
and equations (1) and (2) are satisfied when
$$
\eta = -\frac{1}{3}.
$$
Therefore, for small $\xi_0 \ll 1$, the inner boundary conditions are
$$
\theta(\xi_0) = 1, \qquad \psi(\xi_0) = -\frac{1}{3} \xi_0.
$$

We integrate out to $\xi_1$, determined by the condition
$$
\theta(\xi_1) = 0.
$$
However, an issue with determining $\xi_1$ is that once $\theta(\xi_1)<0$, $\theta^n$ values sometimes become ill-defined for non-integer $n$. To circumvent this issue, we will force our timesteps to decrease when $\theta$ becomes sufficiently small. Specifically, if we force
$$
\Delta \theta = |\theta_{i+1} - \theta_i| < 0.1 |\theta_i|,
$$
for evaluations $\theta_i = \theta(\xi_i)$, then we have that $0.1 |\theta_i| > |(d\theta/d\xi)_i | \Delta \xi = |\psi_i| \Delta \xi$, giving a maximum $\Delta \xi$ of
$$
\Delta \xi_{\rm max} = 0.1 \left| \frac{\theta_i}{\psi_i} \right|.
