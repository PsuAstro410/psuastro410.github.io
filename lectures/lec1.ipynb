## GitHub and Classroom 50

All assignments will be submitted using Classroom 50, which is based on GitHub, but the interface makes sure your assignments are only visible to yourself and the graders (the TA and instructor). 

To use Classroom 50, everyone is required to have a personal GitHub account:

https://docs.github.com/en/account-and-profile/how-tos/account-management/creating-an-account-on-github

**Everyone is required to create a personal account by the end of today, and be added to the Astro 410 Organization**

To install, follow the following steps:

https://github.com/foundation50/classroom50/wiki/Installation

To submit your first assignment (lecture 1, a.k.a this!), follow the directions below:

[Web Guide](https://github.com/foundation50/classroom50/wiki/Web-Student-Guide)

[CLI Guide](https://github.com/foundation50/classroom50/wiki/CLI-Student-Guide)

A few notes on submitting assignments
- Make sure when working on an assignment on your own personal Roar account, you save often. Your connection can get timed out or disconnected, so frequent saving ensures that none of your work is lost
- Committing your assignments often makes sure that you can keep track of what you changed in your homework assignment
- Only submit/push your assignment when you are ready for it to be graded.
