need to become familiar with as we use the language.

The first set of datatypes are similar to those found in other
languages (like C/C++ and Fortran): floating point numbers, integers,
and strings.

The next set of datatypes are containers.  In python, unlike some
languages, these are built into the language and make it very easy to
do complex operations.  We'll look at these later.

Lists group together data.  These are roughly equivalent to the C++ `vector` class.  A list

There are lots of methods that work on lists.  Two of the most useful are append, to add to the end of a list, and pop, to remove the last element:

````python
a.append(6)
a
````

````python
a.pop()
````

### Copying Lists

Copying may seem a little counterintuitive at first.  The best way to think about this is that your list lives in memory somewhere and when you do 

```
a = [1, 2, 3, 4]
```

then the variable `a` is set to point to that location in memory, so it refers to the list.

If we then do
```
b = a
```
then `b` will also point to that same location in memory -- the exact same list object.

Since these are both pointing to the same location in memory, if we change the list through `a`, the change is reflected in `b` as well:

````python
a = [1, 2, 3, 4]
b = a  # both a and b refer to the same list object in memory
print(a)
a[0] = "changed"
print(b)
````

if you want to create a new object in memory that is a copy of another, then you can either index the list, using `:` to get all the elements, or use the `list()` function:

````python
c = list(a)   # you can also do c = a[:], which basically slices the entire list
a[1] = "two"
print(a)
print(c)
````

### Tuples

Tuples are immutable -- they cannot be changed, but they are useful for organizing data in some situations.  We use () to indicate a tuple:

````python
a = (1, 2, 3, 4)
print(a)
````

Tuple unpacking is the primary reason why tuples are used in some instances instead of lists:

````python
w, x, y, z = a
print(w)
print(w, x, y, z)
````

Since a tuple is immutable, we cannot change an element:

````python
a[0] = 2
````

If the tuple is turned into a list, then we can change the data

## Control Flow

To write a program, we need the ability to iterate and take action based on the values of a variable.  This includes if-tests and loops.

**Note:** Python uses whitespace to denote a block of code. A general rule is to use 4 spaces, *not tabs*, to denote a code block

### While loop

A simple while loop&mdash;notice the indentation to denote the block that is part of the loop.

Here we also use the compact `+=` operator: `n += 1` is the same as `n = n + 1`

````python
n = 0
while n < 10:
    print(n)
    n += 1
````

For this type of simple example, we will more often use the `range()` function.  Note that `range()` can take a stride.

````python
for n in range(2, 10, 2):
    print(n)
````

### if statements

`if` allows for branching, and together with `elif`, and `else` it can handle any branching functionality you might need.

````python
x = 0

if x < 0:
    print("negative")
elif x == 0:
    print("zero")
else:
    print("positive")
````

### Iterating over elements

It's easy to loop over items in a list or any _iterable_ object.  The `in` operator is the key here.

````python
alist = [1, 2.0, "three", 4]
for a in alist:
    print(a)
````

````python
for c in "this is a string":
    print(c)
````

We can combine loops and if-tests to do more complex logic, like break out of the loop when you find what you're looking for

````python
n = 0
for a in alist:
    if a == "three":
        break
    else:
        n += 1

print(n)
````

sometimes we want to loop over a list element and know its index -- `enumerate()` helps here:

````python
for n, a in enumerate(alist):
    print(n, a)
````

### List comprehensions

List comprehensions provide a compact way to initialize lists:

````python
squares = [x**2 for x in range(10)]
squares
````

## Basic plotting

plot() is the most basic command.  Here we also see that we can use LaTeX notation for the axes. We first define two numpy arrays:

````python
x = np.linspace(0.0, 2.0*np.pi, 100)
y = np.cos(x)
````

Then plot with the following commands. 

Here, the `ax = fig.add_subplot(111)` is super redundant, because our figure only has one plot. It turns out that the features with `matplotlib.pyplot.subplots` are extremely useful: we will often define a figure with only one subplot, so we can use these features.

````python
fig = plt.figure()
ax = fig.add_subplot(111)

ax.plot(x, y)
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$\cos(x)$")
ax.set_xlim(0, 2*np.pi)
````

we can use symbols instead of lines pretty easily too&mdash;and label them

````python
fig = plt.figure()
ax = fig.add_subplot(111)
ax.plot(x, np.sin(x), "o", label="sine")
ax.plot(x, np.cos(x), "x", label="cosine")
ax.set_xlim(0.0, 2.0*np.pi)
ax.legend()
````

Here we specified the format using a "format string" (see https://matplotlib.org/3.1.1/api/_as_gen/matplotlib.pyplot.plot.html)

This has the form `'[marker][line][color]'`

most functions take a number of optional named arguments too

````python
ax.clear()
ax.plot(x, np.sin(x), linestyle="--", linewidth=3.0)
ax.plot(x, np.cos(x), linestyle="-")
fig
````

### Multiple Axes

There are a wide range of methods for putting multiple axes on a grid.  We'll look at the simplest method.

The `add_subplot()` method we've been using can take 3 numbers: the number of rows, number of columns, and current index

````python
fig = plt.figure()

ax1 = fig.add_subplot(211)

x = np.linspace(0,5,100)
ax1.plot(x, x**3 - 4*x)
ax1.set_xlabel("x")
ax1.set_ylabel(r"$x^3 - 4x$", fontsize="large")

ax2 = fig.add_subplot(212)

ax2.plot(x, np.exp(-x**2))
ax2.set_xlabel("x")
ax2.set_ylabel("Gaussian")

# log scale
ax2.set_yscale("log")

# set the figure size
fig.set_size_inches(6, 8)

# tight_layout() makes sure things don't overlap
fig.tight_layout()
````
