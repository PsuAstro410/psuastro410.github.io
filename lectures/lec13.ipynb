## 4.1 Python Fundamentals

### Lists vs. numpy arrays

A common pattern in this course: compute something in a loop, `append` the results to a **list**, then convert the list to a **numpy array** for math and plotting.

````python
a_list = [1, 2, 3]

# Loop -> list -> array pattern
squares = []
for x in a_list:
    squares.append(x**2)
squares = np.array(squares)
print(squares, type(squares))
````

### `np.linspace` vs. `np.logspace`

- `np.linspace(start, stop, num)`: `num` values **linearly** spaced from `start` to `stop` (endpoint included by default).
- `np.logspace(start, stop, num)`: `num` values **logarithmically** spaced from $10^{\rm start}$ to $10^{\rm stop}$. **The arguments are exponents.**

````python
print(np.linspace(0, 1, num=5))
print(np.logspace(-2, 2, num=5))   # from 10^-2 to 10^2, NOT from -2 to 2
````

### f-strings

f-strings let you put variables directly in strings, with format codes for numbers.

````python
name = "Proxima Centauri"
distance_pc = 1.3020
luminosity = 1.567e-3   # solar luminosities

print(f"{name}: d = {distance_pc:.2f} pc, L = {luminosity:.3e} L_sun")
````

### Functions, default arguments, and `lambda`

A `lambda` is a short, one-line function. It is very useful for passing a function as an input to another function, such as a root finder.

````python
def power_law(x, A, alpha):
    "Return A * x**alpha."
    return A * x**alpha

# The same function, written as a lambda with A and alpha fixed
f = lambda x: 2.0 * x**(-1.5)

print(power_law(4.0, 2.0, -1.5), f(4.0))
````

## 4.2 Units and Physical Constants (cgs)

Convert inputs to **cgs**, do the physics in cgs, and convert the result to convenient units at the end. Never reuse a constant's name (e.g., `G`) as a variable.

````python
G      = 6.674e-8      # cm^3 g^-1 s^-2
M_sun  = 1.989e33      # g
AU     = 1.496e13      # cm
year   = 3.156e7       # s

# Earth's orbital period from Kepler's third law
P = 2 * np.pi * np.sqrt(AU**3 / (G * M_sun))
print(f"P = {P:.4e} s = {P / year:.4f} yr")
````

## 4.3 Classes

A **class** bundles data (**attributes**) and functions that act on that data (**methods**).

- `__init__(self, ...)` runs when you create an **instance**. It stores the inputs as attributes on `self`.
- Methods take `self` as their first argument, so they can use the attributes.
- `__str__(self)` returns the string shown by `print(instance)`.
- A **docstring** documents what the class does and what its attributes mean.

Example: a `Star` class that computes luminosity from the Stefan–Boltzmann law,
$$
L = 4\pi R^2 \sigma_{\rm SB} T_{\rm eff}^4 .
$$

````python
# Constants in cgs
sigma_SB = 5.6704e-5     # erg cm^-2 s^-1 K^-4
R_sun    = 6.957e10      # cm
L_sun    = 3.828e33      # erg/s

class Star:
    '''
    A class to represent a star.

    Attributes
    ----------
    name : str
        Name of the star.
    radius : float
        Stellar radius in solar radii.
    T_eff : float
        Effective temperature in Kelvin.

    Methods
    -------
    luminosity():
        Luminosity in solar luminosities.
    '''

    def __init__(self, name, radius, T_eff):
        self.name   = name      # name of the star
        self.radius = radius    # radius in solar radii
        self.T_eff  = T_eff     # effective temperature in K

    def luminosity(self):
        '''Luminosity from the Stefan-Boltzmann law, in solar luminosities.'''
        radius_cm = self.radius * R_sun                            # convert to cgs
        L_cgs     = 4 * np.pi * radius_cm**2 * sigma_SB * self.T_eff**4   # erg/s
        return L_cgs / L_sun                                       # convert to L_sun

    def __str__(self):
        return f"name={self.name}, R={self.radius} R_sun, T_eff={self.T_eff} K, L={self.luminosity():.3e} L_sun"
````

````python
# Create instances, store them in a list, and loop over the list
sun        = Star("Sun", 1.0, 5772)
sirius_A   = Star("Sirius A", 1.711, 9940)
betelgeuse = Star("Betelgeuse", 764, 3600)

stars = [sun, sirius_A, betelgeuse]

for star in stars:
    print(star)
````

## 4.4 Floating Point and Truncation Errors

Computers store real numbers with finite precision (about 16 significant digits for 64-bit floats).

- **Round-off error:** tiny errors from finite precision.
- **Catastrophic cancellation:** subtracting two nearly equal numbers loses significant digits.
- **Truncation error:** error from approximating a continuous operation (e.g., a derivative) with a finite step.

### Numerical differentiation

Forward difference: $f'(x) \approx \dfrac{f(x+h) - f(x)}{h}$, with truncation error $\propto h$.

Central difference: $f'(x) \approx \dfrac{f(x+h) - f(x-h)}{2h}$, with truncation error $\propto h^2$.

Making $h$ smaller reduces truncation error, but eventually round-off error (cancellation in the numerator) takes over.

## 4.5 Numerical Integration

The **trapezoid rule** approximates the area under a curve with trapezoids:
$$
\int_a^b f(x)\,dx \approx \sum_{i=0}^{N-1} \frac{h}{2}\left[f(x_i) + f(x_{i+1})\right],
$$
with error $\propto h^2$.

````python
def trapezoid(f, a, b, N):
    '''Integrate f from a to b with the trapezoid rule using N intervals.'''
    x = np.linspace(a, b, N + 1)
    h = (b - a) / N
    return h * (0.5 * f(x[0]) + np.sum(f(x[1:-1])) + 0.5 * f(x[-1]))

for N in [4, 16, 64, 256]:
    I = trapezoid(np.sin, 0, np.pi, N)
    print(f"N = {N:4d}: I = {I:.10f}, error = {abs(I - 2):.2e}")
````

Each factor of 4 in $N$ reduces the error by about a factor of 16, as expected for an $h^2$ method.

### Simpson's rule

**Simpson's rule** fits a parabola through each pair of intervals, giving
$$
\int_a^b f(x)\,dx \approx \frac{h}{3}\left[f(x_0) + 4\sum_{i\ {\rm odd}} f(x_i) + 2\sum_{i\ {\rm even}} f(x_i) + f(x_N)\right],
$$
where the sums run over the interior points and $N$ must be **even**. The error is $\propto h^4$.

````python
def simpson(f, a, b, N):
    '''Integrate f from a to b with Simpson's rule using N intervals (N must be even).'''
    if N % 2 != 0:
        raise ValueError("N must be even for Simpson's rule")
    x = np.linspace(a, b, N + 1)
    h = (b - a) / N
    return h / 3 * (f(x[0]) + 4 * np.sum(f(x[1:-1:2])) + 2 * np.sum(f(x[2:-1:2])) + f(x[-1]))

for N in [4, 16, 64, 256]:
    I = simpson(np.sin, 0, np.pi, N)
    print(f"N = {N:4d}: I = {I:.10f}, error = {abs(I - 2):.2e}")
````

Each factor of 4 in $N$ now reduces the error by about a factor of 256, as expected for an $h^4$ method.

## 4.6 Root Finding

We want to solve $f(x) = 0$.

**Bisection:** start with an interval $[a, b]$ where $f(a)$ and $f(b)$ have opposite signs, and repeatedly halve it. Slow but guaranteed to converge.

**Newton–Raphson:** start from a guess $x_0$ and iterate
$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)} .
$$
Very fast near the root (the number of correct digits roughly doubles each step), but needs the derivative $f'(x)$ and a reasonable initial guess. It fails if $f'(x_n) = 0$, and can wander off if the guess is poor.

````python
def newton(f, fprime, x0, tol=1e-10, max_iter=100):
    '''Find a root of f using the Newton-Raphson method, starting from x0.'''
    x = float(x0)
    for i in range(max_iter):
        fx = f(x)
        if abs(fx) < tol:
            return x
        dfx = fprime(x)
        if dfx == 0.0:
            raise RuntimeError(f"Zero derivative at x = {x}")
        x = x - fx / dfx
    raise RuntimeError("Newton-Raphson did not converge")
````

## 4.7 Linear Algebra: Solving $\mathbf{A}\cdot\mathbf{x} = \mathbf{b}$

Many problems reduce to a system of linear equations. In numpy:
- A **matrix** is a 2D array: `np.array([[a11, a12], [a21, a22]])`.
- A **vector** is a 1D array: `np.array([b1, b2])`.
- Matrix multiplication uses `@` (or `np.dot`). The `*` operator is **elementwise**.

### Gaussian elimination with partial pivoting

1. **Forward elimination:** use row operations to make the matrix upper triangular.
2. **Partial pivoting:** before eliminating column $k$, swap in the row with the largest $|A_{ik}|$. This avoids dividing by tiny numbers, which amplifies round-off error.
3. **Back substitution:** solve for $x_n$, then $x_{n-1}$, ..., up to $x_1$.

````python
def gauss_solve(A, b):
    '''Solve A x = b with Gaussian elimination, partial pivoting, and back substitution.'''
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)
    Aug = np.hstack([A, b.reshape(-1, 1)])          # augmented matrix [A | b]

    for k in range(n):
        p = np.argmax(np.abs(Aug[k:, k])) + k        # pivot row
        Aug[[k, p]] = Aug[[p, k]]                    # swap rows
        for i in range(k + 1, n):
            m = Aug[i, k] / Aug[k, k]
            Aug[i, k:] -= m * Aug[k, k:]

    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (Aug[i, -1] - Aug[i, i+1:n] @ x[i+1:]) / Aug[i, i]
    return x
````

### Why pivoting matters

````python
A_bad = np.array([[1e-20, 1.0],
                  [1.0,   1.0]])
b_bad = np.array([1.0, 2.0])

# Without pivoting: eliminate using the tiny 1e-20 pivot
m  = A_bad[1, 0] / A_bad[0, 0]
x2 = (b_bad[1] - m * b_bad[0]) / (A_bad[1, 1] - m * A_bad[0, 1])
x1 = (b_bad[0] - A_bad[0, 1] * x2) / A_bad[0, 0]
print("no pivoting:  ", x1, x2)

print("with pivoting:", gauss_solve(A_bad, b_bad))   # correct answer is x1 =~1, x2 =~1
````

## 4.8 Plotting Checklist

Every plot you turn in should have:

- A figure made with `fig, ax = plt.subplots(1, 1, figsize=(...))`.
- **Axis labels**, with units where relevant: `ax.set_xlabel(r'$t$ [s]')`.
- **Log scales** when data span orders of magnitude: `ax.set_xscale('log')`, `ax.set_yscale('log')`.
- A **legend** when there is more than one curve: `label="..."` in each `ax.plot`, then `ax.legend()`.
- **Equal aspect ratio** when $x$ and $y$ are both lengths, so shapes are not distorted: `ax.set_aspect('equal')`.
- **Markers** for single points, e.g. `ax.plot(x0, y0, 'ko')` for a black dot.

Example: a parametric curve (an Archimedean spiral) with an equal aspect ratio.

````python
theta = np.linspace(0, 6 * np.pi, num=500)
r     = theta / (2 * np.pi)
x     = r * np.cos(theta)
y     = r * np.sin(theta)

fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.plot(x, y, color="C0", label="spiral")
ax.plot(x[-1], y[-1], 'o', color="C1", label="end point")
ax.plot(0, 0, 'ko', label="origin")
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$y$')
ax.set_aspect("equal")
ax.legend()
plt.tight_layout()
plt.show()
````
